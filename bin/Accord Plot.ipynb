```python
import re
import numpy as np
import matplotlib.pyplot as plt

def extract_passive_actor_counts_singleline(filename, actor_id=3):
    counts = []
    # Read entire file as a single string
    with open(filename, 'r') as f:
        text = f.read()
    
    # Match "PassiveActor <id>: ... Count: <numbers>"
    pattern = re.compile(rf"PassiveActor\s*{actor_id}:\s*MolID\s*\d+:\s*Count:\s*([0-9\s]+)")
    matches = pattern.findall(text)
    
    if not matches:
        print(f"Warning: No counts found for PassiveActor {actor_id} in {filename}")
    
    # Combine all matches into a single list of ints
    for match in matches:
        numbers = match.strip().split()
        counts.extend([int(n) for n in numbers])
    
    return np.array(counts)

# -----------------------------
# File paths
# -----------------------------
file1 = "accord_sample_communication_chemical_distance_copy_SEED1 2.txt"
file2 = "accord_sample_communication_chemical_distance_copy_SEED1 12 df.txt"
file3 = "accord_sample_communication_chemical_distance_copy_SEED1 12e-5.txt"

# -----------------------------
# Extract counts
# -----------------------------
counts1 = extract_passive_actor_counts_singleline(file1, actor_id=3)
counts2 = extract_passive_actor_counts_singleline(file2, actor_id=3)
counts3 = extract_passive_actor_counts_singleline(file3, actor_id=3)

# -----------------------------
# Check counts
# -----------------------------
if counts1.size == 0 or counts2.size == 0 or counts3.size == 0:
    raise ValueError("No counts were extracted. Check the actor ID and file formatting!")

# -----------------------------
# Create time axes
# -----------------------------
dt = 0.003  # simulation time step
time1 = np.arange(len(counts1)) * dt
time2 = np.arange(len(counts2)) * dt
time3 = np.arange(len(counts3)) * dt

# -----------------------------
# Apply time offsets
# -----------------------------
offset_2 = time1[-1] + dt
offset_3 = offset_2 + time2[-1] + dt

time2_shifted = time2 + offset_2
time3_shifted = time3 + offset_3

# -----------------------------
# Plot
# -----------------------------
plt.figure(figsize=(12,6))
plt.plot(time1, counts1, label='Result 1 - Passive Actor 3')
plt.plot(time2_shifted, counts2, label='Result 2 - Passive Actor 3')
plt.plot(time3_shifted, counts3, label='Result 3 - Passive Actor 3')
plt.xlabel("Time (s)")
plt.ylabel("Molecule Count")
plt.title("Passive Actor 3 Comparison (Shifted Time)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter1d

# -------------------------------------------------
# Function to extract per-timestep molecule counts
# -------------------------------------------------
def extract_individual_counts(filename, actor_id=3):
    """
    Extract per-timestep (non-cumulative) molecule counts for a given PassiveActor.
    """
    with open(filename, 'r') as f:
        text = f.read()

    # Regex to capture "PassiveActor <id>: ... Count: <numbers>"
    pattern = rf"PassiveActor\s+{actor_id}:.*?Count:\s*((?:\d+\s+)+)"
    matches = re.findall(pattern, text, flags=re.DOTALL)

    if not matches:
        print(f"Warning: No counts found for PassiveActor {actor_id} in {filename}")
        return np.array([])

    # Combine all matches into a single cumulative array
    cumulative_counts = []
    for match in matches:
        nums = [int(x) for x in match.strip().split()]
        cumulative_counts.extend(nums)

    cumulative_counts = np.array(cumulative_counts)

    # Convert cumulative → individual counts
    individual_counts = np.diff(cumulative_counts, prepend=0)
    return individual_counts


# -------------------------------------------------
# File paths
# -------------------------------------------------
file1 = "accord_sample_communication_chemical_distance_copy_SEED1 2.txt"
file2 = "accord_sample_communication_chemical_distance_copy_SEED1 12 df.txt"
file3 = "accord_sample_communication_chemical_distance_copy_SEED1 12e-5.txt"


# -------------------------------------------------
# Extract individual counts
# -------------------------------------------------
counts1 = extract_individual_counts(file1, actor_id=3)
counts2 = extract_individual_counts(file2, actor_id=3)
counts3 = extract_individual_counts(file3, actor_id=3)

# Safety check
if counts1.size == 0 or counts2.size == 0 or counts3.size == 0:
    raise ValueError("No counts were extracted. Check the actor ID and file formatting!")


# -------------------------------------------------
# Create time axes
# -------------------------------------------------
dt = 0.003  # simulation timestep (s)
time1 = np.arange(len(counts1)) * dt
time2 = np.arange(len(counts2)) * dt
time3 = np.arange(len(counts3)) * dt

# -------------------------------------------------
# Apply time offsets so they appear sequentially
# -------------------------------------------------
offset_2 = time1[-1] + dt
offset_3 = offset_2 + time2[-1] + dt

time2_shifted = time2 + offset_2
time3_shifted = time3 + offset_3


# -------------------------------------------------
# Optional: Spread the time axis for better visibility
# -------------------------------------------------
stretch_factor = 5  # Increase to spread horizontally
time1 *= stretch_factor
time2_shifted *= stretch_factor
time3_shifted *= stretch_factor


# -------------------------------------------------
# Smooth the data (Gaussian filter)
# -------------------------------------------------
sigma = 3  # Adjust for smoother or sharper curves
counts1_smooth = gaussian_filter1d(counts1, sigma=sigma)
counts2_smooth = gaussian_filter1d(counts2, sigma=sigma)
counts3_smooth = gaussian_filter1d(counts3, sigma=sigma)


# -------------------------------------------------
# Plot both raw + smoothed curves
# -------------------------------------------------
plt.figure(figsize=(16, 6))

# Raw data (light lines)
plt.plot(time1, counts1, color='blue', alpha=0.3)
plt.plot(time2_shifted, counts2, color='green', alpha=0.3)
plt.plot(time3_shifted, counts3, color='red', alpha=0.3)

# Smoothed data (bold lines)
plt.plot(time1, counts1_smooth, color='blue', linewidth=2.0, label='Result 1 - Smoothed')
plt.plot(time2_shifted, counts2_smooth, color='green', linewidth=2.0, label='Result 2 - Smoothed')
plt.plot(time3_shifted, counts3_smooth, color='red', linewidth=2.0, label='Result 3 - Smoothed')

# Labels and formatting
plt.xlabel("Time (s)")
plt.ylabel("Individual molecule count per timestep")
plt.title("Passive Actor 3 Comparison (Smoothed, Spread Time, Non-cumulative)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
